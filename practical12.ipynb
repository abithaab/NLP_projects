```python
from collections import deque

graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}
```

```python
def bfs(start):
    visited = []
    queue = deque([start])
    
    while queue:
        node = queue.popleft()
        if node not in visited:
            print(node, end="")
            visited.append(node)
            queue.extend(graph[node])
            print("BFS traversal:")
            bfs('A')
```
