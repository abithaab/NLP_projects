```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
import heapq

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')

text = """This is some sample text for summarization. It contains multiple sentences that will be processed to extract key information. The goal is to demonstrate the text summarizer.

Another paragraph for the summarizer. We want to make sure it handles different parts of the input correctly. This part also adds more sentences to the overall text."""

stop_words = set(stopwords.words("english"))

words = word_tokenize(text.lower())


word_frequencies = {}

for word in words:
    if word not in stop_words and word.isalpha():
        word_frequencies[word] = word_frequencies.get(word, 0) + 1


sentences = sent_tokenize(text)
sentence_scores = {}

for sent in sentences:
    for word in word_tokenize(sent.lower()):
        if word in word_frequencies:
            sentence_scores[sent] = sentence_scores.get(sent, 0) + word_frequencies[word]


summary_sentences = heapq.nlargest(3, sentence_scores, key=sentence_scores.get)

summary = " ".join(summary_sentences)

print("Summary:\n")
print(summary)
```
